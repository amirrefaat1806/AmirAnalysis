# 1.Setting up the master folder

```python
from pathlib import Path 
#Path represents file system paths as objects, so you can use / to join paths, .iterdir() to list contents, .is_dir() to check types, and so on.

import json 
#reads and writes JSON files

master_folder = Path(r"O:\archive\projects\2020_euSNN\Neuropixels_RT_mPFC\data\2_processed")
```

# 2. Finding valid sessions and building a summary table

```python

#find folders that contain a share folder and a session_info.json file within it


folders_with_share = [] 
#Creates an empty list. It will collect the session folders that pass the check.


for folder in master_folder.iterdir():
#Loops over every item in the master folder, one at a time, storing each in folder.

    if not folder.is_dir():
        continue
    share_folder = folder / "share"
    session_info_file = share_folder / "session_info.json"
    if share_folder.is_dir() and session_info_file.is_file():
        folders_with_share.append(folder)
i = 0

for folder in folders_with_share:

    current_folder = folders_with_share[i] / Path(r"share\session_info.json")
    i += 1
    with open(current_folder, 'r') as f:
        session_info = json.load(f)

    import pandas as pd

rows = []

for folder in folders_with_share:
    session_info_file = folder / "share" / "session_info.json"

    with open(session_info_file, "r") as f:
        session_info = json.load(f)

    rows.append({
        "session_id": session_info["session_id"],
        "duration_min": session_info["ephys"]["duration_s"] / 60,
        "n_trials": session_info["behaviour"]["n_trials"],
        "n_correct": session_info["behaviour"]["n_correct"],
        "n_early": session_info["behaviour"]["n_early"],
        "n_late": session_info["behaviour"]["n_late"],
        "n_samples": session_info["ephys"]["n_samples"],
    })

Sessions_info = pd.DataFrame(rows)

Sessions_info
```
